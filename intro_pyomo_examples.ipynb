# Introductory Examples on Pyomo

## Example 1:

$$
\boxed{
\begin{aligned}
\max_{x_1,x_2}\quad
&30x_1+40x_2\\
\text{s.t.}\quad
&3x_1+x_2\leq180\\
&2x_1+3x_2\leq240\\
&x_1,x_2\geq0.
\end{aligned}
}
$$

```python
import pyomo.environ as pyo
import pandas as pd
import numpy as np

# Create the optimization model
model = pyo.ConcreteModel()

# Decision variables
model.x1 = pyo.Var(domain=pyo.NonNegativeReals)
model.x2 = pyo.Var(domain=pyo.NonNegativeReals)

# Objective Function
model.obj = pyo.Objective(expr=30*model.x1 + 40*model.x2,
    sense=pyo.maximize)

# Constraints
model.con1 = pyo.Constraint(expr=3*model.x1 + model.x2 <= 180)
model.con2 = pyo.Constraint(expr=2*model.x1 + 3*model.x2 <= 240)

solver = pyo.SolverFactory("glpk")
results = solver.solve(model, tee=False)

print("Solver status:", results.solver.status)
print("Termination condition:", results.solver.termination_condition)

print("Optimal solution:")
print(f"x1 = {pyo.value(model.x1)}")
print(f"x2 = {pyo.value(model.x2)}")
```

## Example 2:

$$
\boxed{
\begin{aligned}
\min_{D,L}\quad
& 1200 \left( 6DL+2L^2\right) \\
\text{s.t.} \quad &2DL^2=8 \\
& D,L \geq 0.
\end{aligned}
}
$$

```python
# Create the optimization model
model = pyo.ConcreteModel()

# Decision variables
model.D = pyo.Var(domain=pyo.NonNegativeReals)
model.L = pyo.Var(domain=pyo.NonNegativeReals)

# Objective Function
model.obj = pyo.Objective(expr=1200*(6*model.D*model.L + 2*model.L**2),
    sense=pyo.minimize)

# Constraints
model.con = pyo.Constraint(expr=2*model.D*model.L**2 == 8)

solver = pyo.SolverFactory("ipopt")
results = solver.solve(model, tee=False)

print("Solver status:", results.solver.status)
print("Termination condition:", results.solver.termination_condition)

print("Optimal solution:")
print(f"D = {pyo.value(model.D)}")
print(f"L = {pyo.value(model.L)}")
```
